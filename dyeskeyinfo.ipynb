```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt
import pandas as pd
# uncomment the following line if using colab
# %pip install lmfit
import lmfit
```

```python
h = 6.626E-34 # kgm^2/s  (Planck's Constant)
m = 9.11E-31 # kg  (mass of an electron)
c = 3.00E8 # m/s (speed of light)
```

```python
dyes = np.array(["1,1'-diethyl-2,2'-cyanine iodide", "1,1'-diethyl-2,2'-carbocyanine iodide", "1,1'-diethyl-2,2'-dicarbocyanine iodide"])
dyes = np.append(dyes, ["1,1'-diethyl-4,4'-cyanine iodide", "1,1'-diethyl-4,4'-carbocyanine iodide", "1,1'-diethyl-4,4'-dicarbocyanine iodide"])
lambda_max = np.array([524, 603, 711])
lambda_max = np.append(lambda_max, [592, 710, 818])
electrons = np.array([6, 8, 10])
electrons = np.append(electrons, [10, 12, 14])

# thiacyanine dyes
dyes = np.append(dyes, ["3,3'-diethylthiacyanine iodide", "3,3'-diethylthiacarbocyanine iodide", "3,3'-diethylthiadicarbocyanine iodide", "3,3'-diethylthiatricarbocyanine iodide"])
lambda_max = np.append(lambda_max, [424, 559, 656, 763])
electrons = np.append(electrons, [6, 8, 10, 12])

bonds = electrons - 2
carbons = electrons - 3
electronpairs = (electrons / 2).astype(int)

boxlength = np.sqrt(lambda_max * 1e-9 * h * (electrons + 1) / (8 * m * c))

boxlength = boxlength * 1e10 # convert to Angstrom

# Build pandas table
df = pd.DataFrame({
    "Dye": dyes,
    "λmax (nm)": lambda_max,
    "Electrons": electrons,
    "Bonds": bonds,
    "Carbons": carbons,
    "Electron Pairs": electronpairs,
    "Box Length (Å)": boxlength
})



display(df)
df.to_csv("dye_data.csv", index=False)
```
